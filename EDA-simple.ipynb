```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('data/dataset.csv')

sns.set_theme(style="whitegrid")
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sns.countplot(data=df, x='cbwd', hue='cbwd', ax=axes[0], palette='Blues_d', order=df['cbwd'].value_counts().index, legend=False)
axes[0].set_title('Frecuencia por Dirección del Viento', fontsize=12)
axes[0].set_xlabel('Dirección del Viento')
axes[0].set_ylabel('Número de Registros')

sns.boxplot(data=df, x='cbwd', y='pm2.5', hue='cbwd', ax=axes[1], palette='Set2', legend=False)
axes[1].set_title('Concentración de PM2.5 según el Viento', fontsize=12)
axes[1].set_xlabel('Dirección del Viento')
axes[1].set_ylabel('PM2.5 (µg/m³)')

plt.tight_layout()
plt.show()
```

```python
nulls_df = pd.DataFrame({
    'Valores Nulos': df.isnull().sum(),
    'Porcentaje (%)': (df.isnull().sum() / len(df)) * 100
})

# Mostrar la tabla completa organizada
display(nulls_df)
```

```python
sns.set_theme(style="whitegrid")
plt.figure(figsize=(10, 4))

sns.boxplot(data=df, x='month', y='pm2.5', hue='month', palette='viridis', legend=False)
plt.title('Distribución de PM2.5 por Mes (Estacionalidad Anual)', fontsize=12)
plt.xlabel('Mes')
plt.ylabel('PM2.5 (µg/m³)')

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 4))

sns.scatterplot(data=df, x='Iws', y='pm2.5', hue='TEMP', palette='coolwarm', alpha=0.5, s=15)
plt.title('Relación Velocidad del Viento vs PM2.5 (Detección de Anomalías por Calma de Viento)', fontsize=12)
plt.xlabel('Suma de Velocidad del Viento (Iws)')
plt.ylabel('PM2.5 (µg/m³)')
plt.legend(title='Temp (°C)', bbox_to_anchor=(1.05, 1), loc='upper left')

plt.tight_layout()
plt.show()
```
